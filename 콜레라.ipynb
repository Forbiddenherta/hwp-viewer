```python
import numpy as np
import pandas as pd
```

```python
cl=pd.read_csv('콜레라 환자.csv') # 콜레라 파일
pu=pd.read_csv('펌프 위치.csv') # 펌프 파일
```

```python
pu_arr=pu[['위도','경도']].to_numpy()
pu_arr # 펌프 배열
cl[['위도','경도']].values.shape#콜레라 환자
```

```python
from scipy.spatial.distance import cdist
```

```python
pu_d=cdist(cl[['위도','경도']].values,pu_arr)
pu_d.shape
```

```python
#가까운 펌프 번호
min_number = np.argmin(pu_d,axis = 1)
```

```python
for i in range (len(pu)) :
    환자수 = np.sum(min_number == i)
    print(f'펌프{i+1} : {환자수}명')
```
